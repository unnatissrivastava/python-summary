```python
#inheritance
class phone:
    def __init__(self, price, brand, camera):
        print("inside phone constructor")
        self.price = price
        self.brand = brand
        self.camera = camera

class smartphone(phone):
    pass
s = smartphone(20000, "apple", 13)
print(s.brand)

    
```

```python
class parent:
    def __init__(self, num):
        self.__num = num 
    def getnum(self):
        return self.__num
class child(parent):
    def show(self):
        print("this is in child class")
son = child(100)
print(son.getnum())
son.show()
```
