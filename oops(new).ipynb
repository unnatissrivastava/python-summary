```python
class Person:                          # <- Parent class
    def __init__(self, name):
        self.name = name

    def greet(self):
        print(f"Hi, my name is {self.name}")


class Student(Person):                  # <- (Person) likhna hi inheritance hai
    def __init__(self, name, roll_no):
        super().__init__(name)           # <- Parent ka __init__ call kiya
        self.roll_no = roll_no


s = Student("Unnati", 5)
s.greet()              # Output: Hi, my name is Unnati  <- Parent ka method use hua!
print(s.roll_no)         # Output: 5   <- Student ka apna data
```

```python
class vehicle:
    def __init__(self, speed):
        self.speed = speed
    def start(self):
        print("vehicle started")
class car(vehicle):
    pass

c = car(10)
c.start()
print(c.speed)
```
