```python
def factorial(n):
    if n==0 or n==1:
        return 1
    else:
        return n * factorial(n-1)

factorial(10)
```

```python
def fibonacci(n):
    if n <= 1:
        return n
    else:
        return fibonacci(n-1) + fibonacci(n-2)

print(fibonacci(6))
```

```python
def return_sum(func, L):
    result = 0

    for i in L:
        if func(i):
            result = result + i

    return result


L = [11, 14, 21, 23, 56, 78, 45, 29, 28]

# Even numbers
x = lambda x: x % 2 == 0

# Odd numbers
y = lambda x: x % 2 != 0

# Multiples of 3
z = lambda x: x % 3 == 0

print(return_sum(x, L))
print(return_sum(y, L))
print(return_sum(z, L))
```

```python
def fib(n):
    if n==0 or n==1:
        return 1
    else:
        return fib(n-1) + fib(n-2)
    
fib(36)
```
